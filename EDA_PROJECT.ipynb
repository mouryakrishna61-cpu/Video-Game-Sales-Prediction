```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df[[
    "NA_Sales",
    "EU_Sales",
    "JP_Sales",
    "Other_Sales",
    "Global_Sales"
]])
plt.title("Outlier Detection in Video Game Sales")
plt.show()
```

```python
Q1 = df["Global_Sales"].quantile(0.25)
Q3 = df["Global_Sales"].quantile(0.75)
IQR = Q3 - Q1
lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR
outliers = df[
    (df["Global_Sales"] < lower_fence) |
    (df["Global_Sales"] > upper_fence)
]
print("Lower Fence:", lower_fence)
print("Upper Fence:", upper_fence)
print("Number of Outliers:", len(outliers))
```

```python
# Fill missing Year with median
df["Year"] = df["Year"].fillna(df["Year"].median())

# Fill missing Publisher with Unknown
df["Publisher"] = df["Publisher"].fillna("Unknown")

# Check missing values again
print(df.isnull().sum())
```

Univariate Analysis :

```python
# Distribution of Global Video Game Sales
sns.histplot(df["Global_Sales"], bins=50, kde=True)
plt.title("Distribution of Global Video Game Sales")
plt.show()
```

Observation:
- Most video games have relatively low global sales.
- Only a small number of games achieve very high sales.
- The distribution has a long tail extending towards higher sales values.
- The distribution is highly positively skewed (right-skewed).

Conclusion:
The analysis shows that most video games generate relatively low sales, while a small number of blockbuster games achieve exceptionally high global sales.

```python
# Number of Games by Genre
sns.countplot(
    data=df,
    x="Genre",
    order=df["Genre"].value_counts().index
)
plt.title("Number of Games by Genre")
plt.xticks(rotation=45)
plt.show()
```

Observation:
- Action has the highest number of games.
- Sports is the second most common genre.
- Miscellaneous and Role-Playing also have a considerable number of games.
- Puzzle has the lowest number of games.

Conclusion:
Action is the most frequently represented genre in the dataset, whereas Puzzle has the fewest games. This indicates that the dataset contains an uneven distribution of games across genres.

Bivariate Analysis :

```python
# Total Global Sales by Genre
genre_sales = df.groupby("Genre")["Global_Sales"].sum().reset_index()
sns.barplot(data=genre_sales, x="Genre", y="Global_Sales")
plt.title("Total Global Sales by Genre")
plt.xticks(rotation=45)
plt.show()
```

Observation:
- Action generates the highest total global sales.
- Sports is the second-highest-selling genre.
- Shooter and Role-Playing also contribute significantly.
- Strategy and Puzzle have comparatively lower total sales.

Conclusion:
Action games contribute the most to total global sales, followed by Sports games. This indicates that these genres have a strong presence in the overall video game sales market.

```python
# Total Global Sales by Year
year_sales = df.groupby("Year")["Global_Sales"].sum().reset_index()
sns.lineplot(data=year_sales, x="Year", y="Global_Sales")
plt.title("Total Global Sales by Year")
plt.show()
```

Observation:
- Global sales increased considerably from the late 1990s.
- Sales reached their highest point around 2008.
- After 2008, total sales generally declined.
- The chart shows very low sales in the final years.

Conclusion:
The dataset shows that total recorded video game sales peaked around 2008 and declined in subsequent years.

Multivariate Analysis :

```python
# Average Global Sales by Genre and Platform
genre_platform_sales = df.pivot_table(
    values="Global_Sales",
    index="Genre",
    columns="Platform",
    aggfunc="mean"
)
plt.figure(figsize=(14, 6))
sns.heatmap(genre_platform_sales, cmap="YlGnBu")
plt.title("Average Global Sales by Genre and Platform")
plt.show()
```

Observation:
- Average sales differ across gaming genres and platforms.
- Some genre-platform combinations show higher average sales than others.
- Certain combinations have no colour-filled cells because there are no records for those combinations.
- Some combinations, such as Shooter on NES and Adventure on Game Boy, show relatively high average sales.

Conclusion:
The heatmap demonstrates that average game sales vary depending on both genre and platform. It helps identify combinations that perform relatively well in the dataset.

```python
top_5_genres = df.groupby("Genre")["Global_Sales"].sum().nlargest(5).index
top_genre_data = df[df["Genre"].isin(top_5_genres)]
genre_year_sales = top_genre_data.groupby(
    ["Year", "Genre"]
)["Global_Sales"].mean().reset_index()
```

```python
plt.figure(figsize=(12, 6))
sns.lineplot(
    data=genre_year_sales,
    x="Year",
    y="Global_Sales",
    hue="Genre"
)
plt.title("Average Global Sales Trends of Top 5 Genres Over the Years")
plt.xlabel("Release Year")
plt.ylabel("Average Global Sales (Millions)")
plt.legend(
    title="Genre",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)
plt.show()
```

Observation:
- Average sales fluctuate across different genres and years.
- Some genres show significant sales spikes during the 1980s and early 1990s.
- After the mid-1990s, average sales for most genres generally remain lower.
- Shooter and Platform genres show some noticeable peaks.

Conclusion:
The analysis indicates that average global video game sales vary across genres and release years. Certain genres experienced higher average sales during specific periods, while sales patterns became relatively stable in later years.

MACHINE LEARNING PART :

```python
from sklearn.model_selection import train_test_split , GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error , mean_squared_error , r2_score
```

Encoding :

```python
categorical_columns = df.select_dtypes(
    include=["object"]
).columns
print("Categorical Columns:")
print(categorical_columns)
```

```python
print("Unique Platforms:", df["Platform"].nunique())
print("Unique Genres:", df["Genre"].nunique())
print("Unique Publishers:", df["Publisher"].nunique())
```

```python
publisher_counts = df["Publisher"].value_counts()

print("Total Publishers:", len(publisher_counts))

print("\nPublishers appearing fewer than 10 times:")
print((publisher_counts < 10).sum())

print("\nPublishers appearing 10 or more times:")
print((publisher_counts >= 10).sum())
```

```python
publisher_counts = df["Publisher"].value_counts()

df["Publisher"] = df["Publisher"].apply(
    lambda x: x if publisher_counts[x] >= 10 else "Other"
)
```

```python
print("Unique Publishers:", df["Publisher"].nunique())
print(df["Publisher"].value_counts().tail())
```

```python
df_encoded = pd.get_dummies(
    df,
    columns=["Platform", "Genre", "Publisher"],
    drop_first=True,
    dtype=int
)

print(df_encoded.head())
print("Original Shape:", df.shape)
print("Encoded Shape:", df_encoded.shape)
```

```python
print("Encoded Dataset Shape:", df_encoded.shape)

print("\nRemaining Object Columns:")
print(df_encoded.select_dtypes(include=["object"]).columns)

print("\nTotal Encoded Columns:")
print(len(df_encoded.columns))
```

```python
df_encoded
```

Separating x and y :

```python
x = df_encoded.drop(
    columns=[
        "Rank",
        "Name",
        "NA_Sales",
        "EU_Sales",
        "JP_Sales",
        "Other_Sales",
        "Global_Sales"
    ]
)
y = df_encoded["Global_Sales"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y,test_size=0.25,random_state=42)
```

Scaling :

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(random_state=42)
}
```

```python
for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    print(name, "training completed")
```

```python
for name, model in models.items():

    y_pred = model.predict(X_test_scaled)

    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    print("Model:", name)
    print("MAE:", mae)
    print("MSE:", mse)
    print("R2 Score:", r2)
    print("-------------------------")
```

HyperParameter :

```python
# Decision Tree
dt_params = {
    "max_depth": [None, 10, 20],
    "min_samples_split": [2, 5]
}

dt_grid = GridSearchCV(
    DecisionTreeRegressor(random_state=42),
    dt_params,
    cv=5,
    scoring="r2",
    n_jobs=-1
)

dt_grid.fit(X_train_scaled, y_train)

print("Decision Tree Best Parameters:", dt_grid.best_params_)
print("Decision Tree Best Score:", dt_grid.best_score_)
```

```python
rf_params = {
    "n_estimators": [50, 100],
    "max_depth": [10, 20],
    "min_samples_split": [2, 5]
}

rf_grid = GridSearchCV(
    RandomForestRegressor(random_state=42),
    rf_params,
    cv=5,
    scoring="r2",
    n_jobs=-1
)

rf_grid.fit(X_train_scaled, y_train)

print("Random Forest Best Parameters:", rf_grid.best_params_)
print("Random Forest Best Score:", rf_grid.best_score_)
```

Model Comparison :

```python
best_models = {
    "Linear Regression": models["Linear Regression"],
    "Decision Tree": dt_grid.best_estimator_,
    "Random Forest": rf_grid.best_estimator_
}

for name, model in best_models.items():

    y_pred = model.predict(X_test_scaled)

    print("Model:", name)
    print("MAE:", mean_absolute_error(y_test, y_pred))
    print("MSE:", mean_squared_error(y_test, y_pred))
    print("R2 Score:", r2_score(y_test, y_pred))
    print("-------------------------")
```

Training the Model :

```python
best_model = rf_grid.best_estimator_
best_model.fit(X_train_scaled, y_train)
```

Prediction :

```python
y_pred = best_model.predict(X_test_scaled)
```
